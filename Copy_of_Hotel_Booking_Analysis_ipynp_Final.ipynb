```python
import sqlite3
import pandas as pd
import os

# Re-establish the connection to the database
conn = sqlite3.connect('hotel_bookings.db')

# Load the dataframe and preprocess it within this cell for self-containment
try:
    df = pd.read_csv('hotel_booking 2.csv')
except FileNotFoundError:
    print("Error: 'hotel_booking 2.csv' not found. Please ensure the file is uploaded to the Colab environment.")
    # Attempt to use the 'df' from the global scope if it exists and the file was not found locally.
    # This assumes 'df' might have been loaded in a previous cell and is still available.
    # If not, the subsequent operations will likely fail, prompting the user to re-upload.
    if 'df' not in globals():
        print("DataFrame 'df' is also not defined globally. Please upload the file and run the necessary cells.")
        conn.close()
        raise # Re-raise the error to stop execution if df is critical and not found

df.drop_duplicates(inplace=True)
df = df.drop('company', axis=1)
df['children'] = df['children'].fillna(0)
df['country'] = df['country'].fillna('Unknown')
df['agent'] = df['agent'].fillna(0)

# Ensure the 'hotel_booking' table exists by recreating it from the DataFrame 'df'
# This makes the cell self-contained and resilient to 'no such table' errors.
df.to_sql('hotel_booking', conn, if_exists='replace', index=False)

query = """
SELECT customer_type,
COUNT(*) AS total_bookings
FROM hotel_booking
GROUP BY customer_type
ORDER BY total_bookings DESC
"""

df_customer = pd.read_sql(query, conn)

display(df_customer)

# Close the connection after use
conn.close()
```

```python
import sqlite3
import pandas as pd

# Re-establish the connection to the database
conn = sqlite3.connect('hotel_bookings.db')

# Ensure the 'hotel_booking' table exists by recreating it from the DataFrame 'df'
# This makes the cell self-contained and resilient to 'no such table' errors.
df.to_sql('hotel_booking', conn, if_exists='replace', index=False)

query = """
SELECT hotel,
ROUND(AVG(adr),2) AS avg_adr
FROM hotel_booking
GROUP BY hotel
"""

df_adr = pd.read_sql(query, conn)

display(df_adr)

# Close the connection after use
conn.close()
```

```python
df_adr.plot(
    x='hotel',
    y='avg_adr',
    kind='bar'
)

plt.title('Average Daily Rate by Hotel')
plt.show()
```
